```python
from sentence_transformers import SentenceTransformer
import pandas as pd
import numpy as np

# Load your dataset
data = pd.read_excel('result.xlsx')

# Load the multilingual Sentence-BERT model
model = SentenceTransformer('distiluse-base-multilingual-cased-v1')

# Create embeddings for context and question
context_embeddings = model.encode(data['context'].tolist(), convert_to_tensor=True)
question_embeddings = model.encode(data['question'].tolist(), convert_to_tensor=True)

# Convert embeddings to numpy arrays
context_embeddings_np = context_embeddings.cpu().detach().numpy()
question_embeddings_np = question_embeddings.cpu().detach().numpy()

# Save the embeddings to files
np.save('context_embeddings.npy', context_embeddings_np)
np.save('question_embeddings.npy', question_embeddings_np)
```
